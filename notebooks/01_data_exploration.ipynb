# 01 — Data Exploration

Load the official TAPE Remote Homology splits and get a feel for the data
before building any features: how many sequences per split, how many
classes, how long the sequences are, and how imbalanced the classes are.

This notebook doesn't produce any experiment results — it's just a sanity
check on the data we'll be using everywhere else.

```
import sys
sys.path.append("../src")

import numpy as np
import matplotlib.pyplot as plt

from data import load_train_val_test

(train_x, train_y), (val_x, val_y), (test_x, test_y) = load_train_val_test()
```

## Split sizes and number of classes

```
print(f"train: {len(train_x)} sequences, {len(set(train_y))} unique labels")
print(f"valid: {len(val_x)} sequences, {len(set(val_y))} unique labels")
print(f"test:  {len(test_x)} sequences, {len(set(test_y))} unique labels")

# Classes seen in train but not in test (or vice versa) matter for a
# classification task -- worth knowing about up front.
train_classes = set(train_y)
test_classes = set(test_y)
print(f"test classes not seen in train: {len(test_classes - train_classes)}")
```

## Sequence length distribution

```
lengths = [len(seq) for seq in train_x]

plt.figure(figsize=(6, 4))
plt.hist(lengths, bins=50)
plt.xlabel("sequence length")
plt.ylabel("count")
plt.title("Training sequence length distribution")
plt.show()

print(f"min: {min(lengths)}, median: {int(np.median(lengths))}, max: {max(lengths)}")
```

## Class (label) distribution\n\nRemote homology has many classes with very few examples each -- this is exactly why macro-averaged metrics (rather than plain accuracy) matter, and why the low-data setting is especially challenging.

```
from collections import Counter

label_counts = Counter(train_y)
counts_sorted = sorted(label_counts.values(), reverse=True)

plt.figure(figsize=(6, 4))
plt.plot(counts_sorted)
plt.xlabel("class rank")
plt.ylabel("number of training examples")
plt.title("Training examples per class (sorted)")
plt.yscale("log")
plt.show()

print(f"number of classes with only 1 training example: {sum(c == 1 for c in label_counts.values())}")
```
